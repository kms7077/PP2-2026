타입힌트는 힘수의 매개변수와 변환값의 타입을 지정하는 목적으로 사용하며, 엄격한 타입지정으로 인한 프로그레머의 의도를 명확히 전달하기 위함이다. 안써도 되지만 파이썬 13부터는 쓰는걸 권장한다.

```python
def greet(msg : str) -> str:
    print("in greet: ", msg)
    return msg

print(greet("안녕"))
```

```python
import random

rol = 10
col = 10

jr = [".", "#"]
field = []

for rol in range(rol):
    field  += [random.choice(jr)*col]

print(field)
```
