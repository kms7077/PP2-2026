# 다중 상속: 2개 이상의 부모클래스로부터 상속받을 때

```python
class Base1:
    def __init__(self,v1 = 10):
        self.attr1 = v1

    def display1(self):
        print(f"{self.attr1}")   

    def display(self):
        self.display1()    
         
    

class Base2:
    def __init__(self,v2 =20):
        self.attr2 = v2

    def display2(self):
        print(f"{self.attr2}")   

    def display(self):
        self.display2()    
             
        

class C(Base1,Base2):
    def __init__(self,v1,v2,v3):
        Base1.__init__(self,v1)
        Base2.__init__(self,v2)
        self.attr3 = v3

    def print(self):
        print(f"{self.attr1}, {self.sttr2}, {self.attr3}")
        self.display1()
        self.display2()
        self.display(self)
        Base2.display(self)

def test():
    
    
     c = C(v1 =1, v2 = 2, v3 = 3)
     c.print()

     c2 = C(1,2,3)
     c2.print()

if __name__ == "__main__":
     test()
```

# Overiding

```python
import math

class shape:
    def __init__(self,name="도형"):
        self.name = name

    def draw(self):
        print(f"shape: {self.name}을(를) 그립니다.")

    def Get_area(self):
        print(f"Shape: {self.name}의 면적을 구힙니다")

class Circle(shape):
    def __init__(self, radius=1, name="원"):
        super().__init__(name)
        self.radius = radius

    def draw(self):
        print(f"shape: {self.name}을(를) 그립니다.")
    
    def Get_area(self):
        print(f"Shape: {self.name}의 면적을 구힙니다")
        return math.pi * self.radius ** 2

class Rectangle(shape):
    def __init__(self, width= 1, hight = 1, name = "사각형"):
        super().__init__(name)
        self.width = width
        self.hight = hight

    def draw(self):
        print(f"shape: {self.name}을(를) 그립니다.")
        
    def Get_area(self):
        print(f"Shape: {self.name}의 면적을 구힙니다")
        return math.self.width * self.hight ** 2


def test():
    c = Circle()

    print(c)

if __name__ == "__main__":
    test()
```
