# Lab 이미지 파일 복사하기

```python
infile = open("123.png", "rb")
outfile = open("kkk.png", "wb")

while True:
    copy_buffer = infile.read(1024)
    if not copy_buffer:
        break
    outfile.write(copy_buffer)

infile.close()
outfile.close()
print("123.png를 kkk.png로 복시하였습니다. ")
```
