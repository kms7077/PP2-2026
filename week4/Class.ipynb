# *장 객채지향 학습

 - 객채지향이란
 - 클래스란?
 - 클래스의 인스턴스 = 객채

```python
class Counter :
    def __init__(self): # 생성함수
        self.count = 0

    def increment(self):
        self.count += 1

    def get_count(self):
        return self.count

def test_counter():
    myc = Counter()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    for _ in range(10):
        myc.increment()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment()
    print(yourc.get_count())
    yourc.increment(4)
    print(yourc.get_count())
    



if __name__ == "__main__":
    test_counter()
```

# TV 리모컨 클래스

- 데이터는 상관없고 동작만 있음 
- 연관된 TV 를 켜거나 꺼라는 동작만 지원
- TV 상태 표시 추가

```python
class TvRemocon:
    def __init__ (self, tv: MyTelevision):
        self.tv = tv

    def press_on(self):
        self.tv.trun_on()

    def press_off(self):
        self.tv.turn_off()

    def press_display(self):
        self.tv.display_status()

def test_remocon():
    tvl = MyTelevision()
    rcl = TvRemocon(tvl)
    rcl.press_on
    rcl.press_display
    rcl.press_off
    rcl.press_display

if __name__ == "__main__":
    test_remocon()
```
